```python
import cv2
import os
import matplotlib.pyplot as plt
# read the image 
image = cv2.imread('Tumor1.jpg', cv2.IMREAD_GRAYSCALE)
print("Image shape:", image.shape)
print("Image type:", image.dtype)
print("Minimum intensity:", image.min())
print("Maximum intensity:", image.max())

plt.figure(figsize=(12, 10))
plt.imshow(image, cmap="gray", vmin=0, vmax=255)
plt.title("Original CT Image")
plt.axis("off")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# Center of suspected tumor
x = 125
y = 90

# Square size: 50 × 50 pixels
size = 35
half = size // 2

# Calculate top-left corner
x_min = x - half
y_min = y - half

# Display CT image
fig, ax = plt.subplots(figsize=(10, 8))
ax.imshow(image, cmap="gray", vmin=0, vmax=255)

# Draw red square
square = Rectangle(
    (x_min, y_min),
    size,
    size,
    edgecolor="red",
    facecolor="none",
    linewidth=2
)

ax.add_patch(square)

# Mark center with blue dot
ax.plot(x, y, "bo", markersize=5)

ax.set_title("Square ROI Around Suspected Tumor")
plt.show()
```

```python
roi = image[y_min:y_min+size, x_min:x_min+size]
plt.figure(figsize=(8, 4))

plt.hist(roi.ravel(), bins=256, range=(0, 256))

plt.title("Histogram of Selected ROI")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.show()
```

```python
adaptive_mean = cv2.adaptiveThreshold(
    roi,
    255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

adaptive_gaussian = cv2.adaptiveThreshold(
    roi,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].imshow(roi, cmap="gray", vmin=0, vmax=255)
axes[0].set_title("Original ROI")

axes[1].imshow(adaptive_mean, cmap="gray", vmin=0, vmax=255)
axes[1].set_title("Adaptive Mean")

axes[2].imshow(adaptive_gaussian, cmap="gray", vmin=0, vmax=255)
axes[2].set_title("Adaptive Gaussian")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```
